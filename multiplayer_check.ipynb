```python
import retro

# Get a list of all available games
games = retro.data.list_games()

# Filter for games that support 2 players
multiplayer_games = []
for game in games:
    try:
        metadata = retro.data.GameMetadata(game)
        if metadata.players == 2:
            multiplayer_games.append(game)
    except Exception as e:
        continue  # Skip games that don't have proper metadata

# Print the multiplayer games
print("Multiplayer Games in Gym-Retro:")
for game in multiplayer_games:
    print(f"- {game}")
```
